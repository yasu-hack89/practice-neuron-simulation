```python
import matplotlib.pyplot as plt

time=[]
voltage=[]

menbrane_potential=-70.0

for t in range(10):
  menbrane_potential += 2.0
  time.append(t)
  voltage.append(menbrane_potential)
  if menbrane_potential >= -55.0:
    break

plt.plot(time, voltage, marker='o')
plt.axhline(y=-55.0, color='green', linestyle='--', label='threshold')
plt.xlabel('time(s)')
plt.ylabel('voltage(mV)')
plt.legend()
plt.show()
```
